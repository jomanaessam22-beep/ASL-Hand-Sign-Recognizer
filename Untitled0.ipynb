```python
# ==============================================================================
# كود معالجة اسم الموديل المضغوط وتشغيل الكاميرا المباشرة فوراً
# ==============================================================================
import os, io, glob, zipfile, shutil, base64, string
import numpy as np
import tensorflow as tf
from PIL import Image
from IPython.display import display, HTML, JSON
from google.colab import output

# 1. كشف الموديل المرفوع وإصلاح صيغته تلقائياً
model = None
model_path = None

print("🔍 جاري فحص الملف المرفوع وإصلاح الصيغة...")

# البحث عن الملفات الموجودة في الكولاب
all_files = glob.glob("/content/*")

# أ) إذا كان الملف .keras موجوداً بشكل مباشر
for f in all_files:
    if f.endswith(".keras"):
        try:
            model = tf.keras.models.load_model(f, compile=False)
            model_path = f
            break
        except:
            pass

# ب) إذا كان الملف مضغوطاً أو ينتهي بـ .zip (مثل الملف المرفوع لديكِ)
if model is None:
    zip_files = [f for f in all_files if f.endswith(".zip") or ".keras" in f]
    for zf in zip_files:
        if zipfile.is_zipfile(zf):
            with zipfile.ZipFile(zf, 'r') as z:
                names = z.namelist()

                # استخراج ملف labels.txt إن وجد
                if "labels.txt" in names:
                    z.extract("labels.txt", "/content")
                    print("📄 تم استخراج ملف الحروف labels.txt")

                # الحالة 1: الويندوز أضاف .zip للملف والملف داخلياً هو موديل Keras
                if 'config.json' in names:
                    fixed_name = "/content/fixed_asl_model.keras"
                    shutil.copy(zf, fixed_name)
                    model = tf.keras.models.load_model(fixed_name, compile=False)
                    model_path = fixed_name
                    print(f"✅ تم إصلاح امتداد الملف وتحميله بنجاح من: {zf}")
                    break

                # الحالة 2: ملف zip عادي وبداخله ملف .keras
                keras_inside = [n for n in names if n.endswith(".keras")]
                if keras_inside:
                    z.extract(keras_inside[0], "/content")
                    extracted_path = os.path.join("/content", keras_inside[0])
                    model = tf.keras.models.load_model(extracted_path, compile=False)
                    model_path = extracted_path
                    print(f"✅ تم فك الضغط وتحميل الموديل بنجاح: {extracted_path}")
                    break

if model is None:
    raise RuntimeError("لم نتمكن من قراءة الموديل، تأكدي من وجود الملف في مجلد /content")

# 2. تحديد قائمة الحروف وأبعاد الصورة
labels_file = "/content/labels.txt"
if os.path.exists(labels_file):
    class_names = [l.strip().split()[-1] for l in open(labels_file) if l.strip()]
else:
    class_names = list(string.ascii_uppercase)

IMG_SIZE = tuple(model.input_shape[1:3])
print(f"🎉 الموديل جاهز للعمل! أبعاد الصورة: {IMG_SIZE} | عدد الحروف: {len(class_names)}")

# 3. دالة التعرف المباشر على الحرف
def predict_live_frame(img_data_url):
    try:
        header, encoded = img_data_url.split(",", 1)
        img_bytes = base64.b64decode(encoded)
        pil_img = Image.open(io.BytesIO(img_bytes)).convert("RGB")

        arr = np.asarray(pil_img, dtype=np.float32)
        arr = tf.image.resize(arr, IMG_SIZE, method="bilinear").numpy()
        arr = arr[None, ...]

        probs = model.predict(arr, verbose=0)[0]
        top_idx = int(np.argmax(probs))
        top_char = class_names[top_idx]
        top_conf = float(probs[top_idx] * 100)

        order = np.argsort(probs)[::-1][:3]
        top3 = [f"{class_names[i]}: {probs[i]*100:.1f}%" for i in order]

        return JSON({
            "status": "success",
            "char": str(top_char),
            "conf": round(top_conf, 1),
            "top3": " | ".join(top3)
        })
    except Exception as e:
        return JSON({"status": "error", "char": "--", "conf": 0.0, "top3": str(e)})

output.register_callback('predict_live_frame', predict_live_frame)

# 4. واجهة الكاميرا التفاعلية الحية
live_camera_html = """
<div id="asl-container" style="max-width: 520px; margin: 10px auto; font-family: 'Segoe UI', Tahoma, sans-serif; background: #18191a; padding: 20px; border-radius: 16px; box-shadow: 0 8px 24px rgba(0,0,0,0.5); text-align: center; color: white;">

    <div id="asl-banner" style="background: #2e7d32; padding: 12px; border-radius: 10px; margin-bottom: 15px; transition: background 0.3s;">
        <div style="font-size: 14px; opacity: 0.9; margin-bottom: 4px;">الحرف المكتشف حالياً:</div>
        <div id="asl-char" style="font-size: 52px; font-weight: 900; line-height: 1.1; letter-spacing: 2px;">--</div>
        <div id="asl-conf" style="font-size: 16px; font-weight: bold; margin-top: 4px;">نسبة التأكيد: 0%</div>
        <div id="asl-top3" style="font-size: 12px; opacity: 0.8; margin-top: 6px;">في انتظار يدك داخل المربع...</div>
    </div>

    <div style="position: relative; width: 480px; height: 360px; margin: 0 auto; overflow: hidden; border-radius: 12px; border: 2px solid #333;">
        <video id="asl-video" playsinline autoplay muted style="width: 100%; height: 100%; object-fit: cover; transform: scaleX(-1); display: block;"></video>
        <div id="asl-guide" style="position: absolute; width: 250px; height: 250px; left: 115px; top: 55px; border: 4px solid #00e676; border-radius: 12px; box-sizing: border-box; pointer-events: none; box-shadow: 0 0 15px rgba(0,230,118,0.7);">
            <span style="position: absolute; bottom: 8px; left: 0; right: 0; font-size: 13px; color: #00e676; font-weight: bold; background: rgba(0,0,0,0.6); padding: 2px 0;">ضع اليد هنا</span>
        </div>
    </div>

    <button id="asl-stop-btn" style="margin-top: 15px; width: 100%; padding: 12px; background: #e53935; color: white; border: none; border-radius: 8px; font-size: 16px; font-weight: bold; cursor: pointer;">
        ⏹ إيقاف الكاميرا
    </button>
</div>

<script>
(async function() {
    if (window.activeASLStream) window.activeASLStream.getTracks().forEach(t => t.stop());
    if (window.activeASLTimer) clearInterval(window.activeASLTimer);

    const video = document.getElementById('asl-video');
    const charEl = document.getElementById('asl-char');
    const confEl = document.getElementById('asl-conf');
    const top3El = document.getElementById('asl-top3');
    const banner = document.getElementById('asl-banner');
    const stopBtn = document.getElementById('asl-stop-btn');

    try {
        const stream = await navigator.mediaDevices.getUserMedia({
            video: { width: { ideal: 640 }, height: { ideal: 480 }, facingMode: 'user' }
        });
        window.activeASLStream = stream;
        video.srcObject = stream;
        await video.play();
    } catch(err) {
        charEl.innerText = "خطأ";
        confEl.innerText = "تعذر فتح الكاميرا: " + err.message;
        banner.style.background = "#c62828";
        return;
    }

    const canvas = document.createElement('canvas');
    canvas.width = 128;
    canvas.height = 128;
    const ctx = canvas.getContext('2d');
    let isProcessing = false;

    window.activeASLTimer = setInterval(async () => {
        if (isProcessing || !video.videoWidth || !video.videoHeight) return;
        isProcessing = true;
        try {
            const vw = video.videoWidth, vh = video.videoHeight;
            const s = Math.min(vw, vh) * 0.72;
            const sx = (vw - s) / 2, sy = (vh - s) / 2;

            ctx.save();
            ctx.translate(canvas.width, 0);
            ctx.scale(-1, 1);
            ctx.drawImage(video, sx, sy, s, s, 0, 0, canvas.width, canvas.height);
            ctx.restore();

            const dataUrl = canvas.toDataURL('image/jpeg', 0.85);
            const res = await google.colab.kernel.invokeFunction('predict_live_frame', [dataUrl], {});
            const data = res.data['application/json'];

            if (data && data.status === "success") {
                charEl.innerText = data.char;
                confEl.innerText = `نسبة التأكيد: ${data.conf}%`;
                top3El.innerText = data.top3;

                if (data.conf >= 75) banner.style.background = "#2e7d32";
                else if (data.conf >= 45) banner.style.background = "#f57c00";
                else banner.style.background = "#546e7a";
            }
        } catch(e) {}
        finally { isProcessing = false; }
    }, 250);

    stopBtn.onclick = () => {
        if (window.activeASLTimer) clearInterval(window.activeASLTimer);
        if (window.activeASLStream) window.activeASLStream.getTracks().forEach(t => t.stop());
        const container = document.getElementById('asl-container');
        if (container) container.remove();
    };
})();
</script>
"""

display(HTML(live_camera_html))
```
